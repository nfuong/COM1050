```python
toan, ly, hoa, anh = map(float, input("Nhập điểm: ").split())
tb = (toan + ly + hoa + anh) / 4
liet = (toan < 4.0 or ly < 4.0 or hoa < 4.0 or anh <4.0)
if liet:
  print('Học lực kém')
elif tb >= 9.0:
    print('Xuất sắc')
elif tb >= 8.5:
    print('Giỏi')
elif tb >= 7.0:
    print('Khá')
elif tb >= 5.5:
    print('Trung bình')
else:
    print('Yếu')
print(f"Điểm trung bình = {tb:.2f}")
```

```python
h, w = map(float,input("Nhập chiều cao (m) và cân nặng (kg): ").split())
BMI = w/h**2
print(f"BMI của bạn = {BMI:.2f}")
if BMI < 18.5:
  print('Thiếu cân')
elif BMI < 25:
  print('Bình thường')
elif BMI < 30:
  print('Thừa cân')
else:
  print('Béo phì')
```
